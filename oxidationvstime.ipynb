```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 读取同目录下由 C++ 写出的 oxidegrow.csv
csv_file = "oxidegrow.csv"

df = pd.read_csv(csv_file)

# 看一下前几行，确认列名和数据是否正常
df.head()
```

```python
df.columns
```

```python
plt.figure(figsize=(6, 4))

plt.plot(df["time"], df["Total_oxide"])
plt.xlabel("Time (ms)")
plt.ylabel("Total Oxide Count")
plt.title("Total Oxide Growth vs Time")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.plot(df["time"], df["Si_oxide"],   label="Si oxide")
plt.plot(df["time"], df["Cr_oxide"],   label="Cr oxide")
plt.plot(df["time"], df["Total_oxide"], label="Total oxide")

plt.xlabel("Time (ms)")
plt.ylabel("Oxide Count")
plt.title("Oxide Growth vs Time")
plt.legend()
plt.grid(True)

plt.show()
```

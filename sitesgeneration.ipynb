```python
import numpy as np
import pandas as pd

# ======================
# 参数
# ======================
box_size = 1e-3  # 立方体边长

n_type0 = 100
n_type1 = 1800
n_type3 = 1000

types = (
    [0] * n_type0 +
    [1] * n_type1 +
    [3] * n_type3
)

n_total = len(types)

# ======================
# 生成随机坐标
# ======================
coords = np.random.rand(n_total, 3) * box_size  # x, y, z ∈ [0, box_size)

# ======================
# 组装 DataFrame（不打散）
# ======================
df = pd.DataFrame(coords, columns=["x", "y", "z"])
df["type"] = types
df.insert(0, "id", np.arange(n_total))  # id: 0 ~ n_total-1

# ======================
# 输出 CSV
# ======================
out_file = "sites3type.csv"
df.to_csv(out_file, index=False)

# 检查
print(df["type"].value_counts())
print(df.head(10))
print(df.tail(10))
```

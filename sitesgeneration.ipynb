```python
import numpy as np
import pandas as pd

# 参数
box_size = 1e-3      # 立方体边长
n_per_type = 1000    # 每种 type 的点数
```

```python
n_total = 2 * n_per_type

# 在 [0, box_size] 内均匀随机生成坐标
coords = np.random.rand(n_total, 3) * box_size  # 列顺序为 x, y, z

# 生成 type：前一半是 0，后一半是 1
types = np.array([0] * n_per_type + [1] * n_per_type)

# 如果想让 0 和 1 在空间里打散，而不是前 1000 行全 0、后 1000 行全 1，可以打乱顺序：
# idx = np.arange(n_total)
# np.random.shuffle(idx)
# coords = coords[idx]
# types = types[idx]

# 组装成 DataFrame
df = pd.DataFrame(coords, columns=["x", "y", "z"])
df["type"] = types
df.insert(0, "id", np.arange(n_total))  # 在最前面插入 id 列（0 到 n_total-1）

# 保存为 CSV
out_file = "sites2type.csv"
df.to_csv(out_file, index=False)

# 看前几行确认一下
df.head()
```
